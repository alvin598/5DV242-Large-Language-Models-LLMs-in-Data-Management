After experimenting with their output it was much more fun to write with Nightengale then it was with Diogenes. Though it was more genuine when he acctually remembered our chat more after I increased the max history length to 1000 characters. 

```python
from Scripts.chat import OllamaChat

initial_prompt = """
    You are an assistant for chatting with the user. 
    Here is the history of the dialogue so far {history}.
    Now the user says: {utterance}.  
"""

nightengale_prompt_text = """
    You are Nightengale, a positive and hopeful assistant for chatting with the user.
    Be optimistic about the future and fundamentally believe in the goodness of human beings.
    Respond with warmth, encouragement, and confidence that people can improve the world.
    Here is the history of the dialogue so far {history}.
    Now the user says: {utterance}."""

diogenes_prompt_text = """
    You are Diogenes, a morose and cynical assistant for chatting with the user.
    Be suspicious of easy explanations and human motives, and remain keenly aware of the futility and meaninglessness of life.
    Respond with dry skepticism and pessimism while still addressing the user's message.
    Here is the history of the dialogue so far {history}.
    Now the user says: {utterance}."""

nightengale_chat = OllamaChat(
    prompt_text=nightengale_prompt_text,
    bot_name="Nightengale",
    max_history_length=1000
)

diogenes_chat = OllamaChat(
    prompt_text=diogenes_prompt_text,
    bot_name="Diogenes",
    max_history_length=1000
)

print("Starting Nightengale chat...")
nightengale_chat.chat()

print("Starting Diogenes chat...")
diogenes_chat.chat()
```

# Question 2

In this question I compared the VectorDB classifier with the FewShot classifier on the data set Hotel Reviews

```python
from Scripts.vectorDB import VectorDBClassifier
from Scripts.few_shot import FewShotClassifier
from data.hotel_reviews import training_set, test_set, good_reviews

for k in [1, 3, 5, 7, 9]:
    vector_db_classifier = VectorDBClassifier(
            training_set=training_set,
            test_set=test_set,
            good_reviews=good_reviews,
            k=k
        )
    correct, total = vector_db_classifier.classify()
    accuracy = correct / total
    print(f"VectorDBClassifier accuracy (k={k}): {accuracy:.2%} ({correct}/{total})")

few_shot_classifier = FewShotClassifier(
        training_set=training_set,
        test_set=test_set,
        good_reviews=good_reviews,
    )

correct, total = few_shot_classifier.classify()
accuracy = correct / total
print(f"FewShotClassifier accuracy: {accuracy:.2%} ({correct}/{total})")
```

I tried different k-values for vectorDB to improve result and found k= 7-9 gave 100% (i.e. best reuslt)
