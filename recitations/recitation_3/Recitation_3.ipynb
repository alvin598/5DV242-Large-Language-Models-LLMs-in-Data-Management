# Question 2

In this question I compared the VectorDB classifier with the FewShot classifier on the data set Hotel Reviews

```python
from scripts.vectorDB import VectorDBClassifier
from data.hotel_reviews import good_reviews, bad_reviews
import random

accuracies = []

for run in range(10):
    questions = good_reviews + bad_reviews
    random.shuffle(questions)
    training_set = questions[:80]
    test_set = questions[80:]

    vector_db_classifier = VectorDBClassifier(
        training_set=training_set,
        test_set=test_set,
        good_reviews=good_reviews,
        k=9
    )
    correct, total = vector_db_classifier.classify()
    accuracy = correct / total
    accuracies.append(accuracy)
    print(f"Run {run + 1}: {correct}/{total} ({accuracy:.1%})")

print(f"Average accuracy: {sum(accuracies) / len(accuracies):.1%}")
```
